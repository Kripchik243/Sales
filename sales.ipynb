```python
import pandas as pd

df = pd.read_csv("sales_sample.csv", sep=",")
counts_product = df.groupby("product")["quantity"].sum()
counts_product.plot(kind="bar", xlabel="Вид продукта", ylabel="Количество проданных единиц")
print(counts_product)
```

```python
counts_city = df.groupby("region")["revenue_rub"].sum()
counts_city.plot(kind="bar", xlabel="Регион", ylabel="Выручка")
print(counts_city)
```

```python
counts_market = df["sales_channel"].value_counts()
counts_market.plot(kind="bar", xlabel="Канал продаж", ylabel="Количество заказов")
print(counts_market)
```

```python
avg_by_city = df.groupby("region")["revenue_rub"].mean()
avg_by_city.plot(kind="bar", xlabel="Города", ylabel="Средняя сумма заказа до скидки")
print(avg_by_city)
```
