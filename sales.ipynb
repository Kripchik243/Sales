```python
import pandas as pd

df = pd.read_csv("sales_sample.csv", sep=",")
print(df.columns)
counts_product = df["product"].value_counts()
counts_product.plot(kind="bar", xlabel="Вид продукта", ylabel="Количество продаж")
print(counts_product)
```

```python
counts_city = df["region"].value_counts()
counts_city.plot(kind="bar", xlabel="Регион", ylabel="Количество продаж")
print(counts_city)
```

```python
counts_market = df["sales_channel"].value_counts()
counts_market.plot(kind="bar", xlabel="Маркетплейс", ylabel="Количество заказов")
print(counts_market)
```

```python
avg_by_city = df.groupby("region")["revenue_rub"].mean()
avg_by_city.plot(kind="bar", xlabel="Города", ylabel="Средний чек")
print(avg_by_city)
```
